```python
import numpy as np
from sklearn.svm import SVC
X = np.array([
    [1,1],
    [0,1],
    [2,2],
    [3,2],
    [2,3]
    ])
y = np.array([-1,-1,1,1,1])
```

```python
clf = SVC(kernel='linear')
clf.fit(X,y)
```

```python
print("W coef ",clf.coef_)
print("b coef ",clf.intercept_)
```

```python
print("Support Vectors ",clf.support_vectors_)
```

```python
print("Support vector Indices ",clf.support_)
```

```python
clf.predict(X)
```

```python
clf.score(X,y)
```

```python
print(2/np.linalg.norm(clf.coef_[0]))
```

```python
#with plotting
import numpy as np
from sklearn.svm import SVC
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions
X = np.array([[1,1],[0,1],[2,2],[3,2],[2,3]],dtype="float")
y = np.array([-1,-1,1,1,1])
```

```python
y_plot = (y == 1).astype(int)
clf = SVC(kernel='linear').fit(X,y_plot)
```

```python
plot_decision_regions(X,y_plot,clf=clf,X_highlight=clf.support_vectors_,legend=2)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("SVM Decision Boundaries")
plt.show()
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
```

```python
est = []
est.append(("KNN",KNeighborsClassifier(n_neighbors=5)))
est.append(("LR",LogisticRegression(max_iter=300)))
est.append(("SVC",SVC(kernel="linear")))
```

```python
est
```

```python
voting = VotingClassifier(estimators=est,voting="hard")
```

```python
voting.fit(X_train,y_train)
```

```python
voting.score(X_train,y_train)
```

```python
voting.score(X_test,y_test)
```
