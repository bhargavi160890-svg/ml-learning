```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://free-proxy-list.net/en/"
```

```python
with requests.Session() as session:
    proxies_page = session.get(url)
proxies_page
```

```python
soup = BeautifulSoup(proxies_page.content,'html.parser')
soup
```

```python
#table table-striped table-bordered
proxies_table = soup.find('table',class_='table table-striped table-bordered')
proxies_table
```

```python
proxies = []
for row in proxies_table.tbody.find_all('tr'):
  proxies.append('{}:{}'.format(row.find_all('td')[0].string,row.find_all('td')[1].string))
```

```python
proxies
```

```python
columns = [th.text.strip() for th in proxies_table.find_all('th')]
columns
```

```python
rows = []
for row in proxies_table.tbody.find_all('tr'):
  row_data = [td.text.strip() for td in row.find_all('td')]
  rows.append(row_data)
```

```python
rows
```

```python
import csv
csv_file = 'proxies.csv'
with open(csv_file,'w',newline='') as file:
  writer = csv.writer(file)
  writer.writerow(columns)
  writer.writerows(rows)
```
