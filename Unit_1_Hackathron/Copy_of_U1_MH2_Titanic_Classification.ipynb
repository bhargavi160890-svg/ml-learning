### 3.4 Working on the "Age" column (2 Marks)
find the number of NaN entries in the age column and their row index. Calculate the mean, Standard deviation of the Age column and check the distribution of the age column.We can fill the missing values with randomly generated integer values between (mean+Standard deviation, mean-Standard deviation). Use : np.isnan; np.random.randint; concept of slicing dataframe. Convert the age column as an integer data type.



```
# wrongindex_anull=df['Age'].isna().index
def impute_age(df1):
  np.random.seed(42) #ensure same random Age Values every run
  index_anull=df1[df1['Age'].isna()].index
  ms_sum=np.mean(df1['Age'])+np.std(df1['Age'])
  ms_sub=np.mean(df1['Age'])-np.std(df1['Age'])
  for i in index_anull:
    df1.loc[i,'Age']=np.random.randint(ms_sub,ms_sum)
  df1['Age']=df1['Age'].astype(int)
  return df1
df_train=impute_age(df_train)
df_train['Age'].isna().sum()
```

### 3.10 Create a single function for preprocessing the test set (X_test) and apply it. (4 Marks)
#### **Note**: All the pre-processing steps that were applied on the train set before ML Modelling are also applied on the test set before passing through the predict function.

```
## Create a function
def preprocess(df_test):
  df_test['Has_cabin'] = df_test['Cabin'].notna().astype(int)
  df_test['family_size'] = df_test['SibSp'] + df_test['Parch']+1
  df_test['Embarked']=df_test['Embarked'].map({'S':1,'C':2,'Q':3}).fillna(1).astype(int)
  df_test=impute_age(df_test)
  df_test['Sex']=df_test['Sex'].map({'male':1,'female':0}).astype(int)
  df_test.drop(columns=['PassengerId', 'Name', 'SibSp', 'Parch', 'Ticket', 'Cabin'],inplace=True,errors='ignore')
  return df_test
```

```
#Step 1 — Define all features
from itertools import combinations
#combinations — the key idea
all_features = ['Pclass', 'Sex', 'Age', 'Fare', 'Embarked', 'Has_cabin', 'familt_size']
best_score = 0
best_features = all_features

# Try dropping 1 or 2 features at a time
for r in [1, 2]:
    for drop_cols in combinations(all_features, r):
        keep_cols = [c for c in all_features if c not in drop_cols] # remaining columns
        idx = [all_features.index(c) for c in keep_cols]  # their column positions (numbers)
        X_sub_train = X_train_scaled[:, idx] #select only those columns
        X_sub_test  = X_test_scaled[:, idx]
        lgs_sub = LogisticRegression(C=10)
        lgs_sub.fit(X_sub_train, y_train)
        score = lgs_sub.score(X_sub_test, y_test)
        if score > best_score:
            best_score = score
            best_features = keep_cols
            best_drop = drop_cols

print(f"Best accuracy : {best_score:.4f}")
print(f"Drop columns  : {best_drop}")
print(f"Keep columns  : {best_features}")
```
