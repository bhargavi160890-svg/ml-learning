```python
import numpy as np
emb = {}
with open("/content/glove.6B.50d.txt",encoding="utf8") as f:
  for line in f:
    parts = line.split()
    emb[parts[0]] = np.array(parts[1:],dtype=np.float32)
print(emb["king"][:5])
print(len(emb["king"]))
```

```python
def cosine(a,b):
  return a @ b /(np.linalg.norm(a)*np.linalg.norm(b))
```

```python
result = emb["king"] -emb["man"] +emb["woman"]
inputs = {"king","man","woman"}
best = max((w for w in emb if w not in inputs),key=lambda w:cosine(result,emb[w]))
print(best)
```

```python
from gensim.models import Word2Vec
sentences =[
            ['the', 'cat', 'sat', 'on', 'mat'],
            ['the', 'dog', 'sat', 'on', 'mat'],
            ['the', 'cat', 'ate', 'fish']
            ]
model = Word2Vec(sentences,vector_size=5,window=2,min_count=1,sg=1,epochs=500)
```

```python
print(model.wv["cat"])
```

```python
print(model.wv.most_similar("cat",topn=1))
```

```python
predictions = model.predict_output_word(context_words_list=["cat","on"],topn=1)
print(predictions)
```

```python
import spacy

# Load English tokenizer, tagger, parser and NER
nlp = spacy.load("en_core_web_sm")
```

```python
text = " john is working on Google in New Gersy"
```

```python
doc = nlp(text)
```

```python
for ent in doc.ents:
  print(ent.text,ent.label_)
```

```python
#Rawtext->tokenize->postag->chunk->parse->NER->MEaning
```

```python
print("Raw text",text)
```

```python
for token in doc:
  print(token.text)
```

```python
for token in doc:
  print(token.text,"->",token.pos_,"(",token.tag_,")")
```

```python
for chunk in  doc.noun_chunks:
  print(chunk.text,"->",chunk.label_)
```

```python
for token in doc:
  print(token.text,"->",token.dep_)
```

```python
for ent in doc.ents:
  print(ent.text,"->",ent.label_)
```

```python
for token in doc:
    if token.dep_ =="nsubj":
        print("Subject:",token.text)
    if token.dep_ =="Root":
        print("Acttion:",token.text)
```
