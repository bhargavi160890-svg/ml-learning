### Chatbot Architecture

Defining functions for Loading Intent, Collecting params, Checking actions, Getting Attributes and Identifying Intents

````python
def loadIntent(path, intent):
    with open(path) as fil:
        dat = json.load(fil)
        intent = dat[intent]
        return Intent(intent['intentname'],intent['Parameters'], intent['actions'])

def check_required_params(current_intent, attributes, context):
    '''Collects attributes pertaining to the current intent'''
    for para in current_intent.params:
        if para.required:
            if para.name not in attributes:
                return random.choice(para.prompts), context
    return None, context

def check_actions(current_intent, attributes, context):
    '''This function performs the action for the intent as mentioned
    in the intent config file. Performs actions pertaining to current intent '''
    context = IntentComplete()
    if current_intent.action.endswith('()'):
        return eval(current_intent.action), context
    return current_intent.action, context

def getattributes(uinput,context,attributes, intent):
    '''This function marks the slots in user input, and updates
    the attributes dictionary'''
    uinput = " "+uinput.lower()+" "
    if context.name.startswith('IntentComplete'):
        return attributes, uinput
    else:
        files = os.listdir(path_slots)
        slots = {}
        for fil in files:
            if fil == ".ipynb_checkpoints":
                continue
            lines = open(path_slots+fil).readlines()
            for i, line in enumerate(lines):
                line = line.strip()
                if len(uinput.split(" "+line.lower()+" ")) > 1:
                    slots[line] = fil[:-4]
        for value, slot in slots.items():
            if intent != None and slot in " ".join([param.name for param in intent.params]):
                uinput = re.sub(value,r'$'+slot,uinput,flags=re.IGNORECASE)
                attributes[slot] = value
            else:
                uinput = re.sub(value,r'$'+slot,uinput,flags=re.IGNORECASE)
                attributes[slot] = value
        return attributes, uinput

def input_processor(user_input, context, attributes, intent):
    '''Update the attributes, abstract over the slots in user input'''
    attributes, cleaned_input = getattributes(user_input, context, attributes, intent)
    return attributes, cleaned_input

def intentIdentifier(clean_input, context,current_intent):
    clean_input = clean_input.lower()
    if (current_intent==None):
        return loadIntent(path_param,intentPredict(clean_input))
    else:
        #If current intent is not none, stick with the ongoing intent
        #return current_intent
        intent = loadIntent(path_param,intentPredict(clean_input))
        if current_intent != intent:
            for para in current_intent.params:
                if para.name in clean_input:
                    return current_intent
        return loadIntent(path_param,intentPredict(clean_input))
````

Session class is one active session of the chatbot which the user interacts with. Let's go into the details:

**reply( )** is the important one in our session object it takes user_input as a parameter and calls different modules of the chatbot architecture:


*   **input_processor( )** - It helps in preprocessing and fetching the slots that can identify in the ready state
    
    - **getattributes( )** - It helps in identifying all the slots in the user utterance. Identify and map them to the parameters
    
    
*   **intentIdentifier( )**

  -  **intentPredict()** - Task to complete

*   **check_required_params( )** - Based on the current intents, it goes over it's parameters

*   **check_actions( )** - This function performs the action for the intent

**Note:** Refer the *Chatbot_Reading_Material.pdf* for more information on the conversation flow


       


````python
class Session:
    def __init__(self, attributes=None, active_contexts=[FirstGreeting(), IntentComplete() ]):
        '''Initialise a default session'''
        # Active contexts not used yet, can use it to have multiple contexts
        self.active_contexts = active_contexts

        # Contexts are flags which control dialogue flow
        self.context = FirstGreeting()

        # Intent tracks the current state of dialogue
        self.current_intent = None

        # attributes hold the information collected over the conversation
        self.attributes = {}

    def reply(self, user_input):
        '''Generate response to user input'''
        self.attributes, clean_input = input_processor(user_input, self.context, self.attributes, self.current_intent)

        self.current_intent = intentIdentifier(clean_input, self.context, self.current_intent)

        prompt, self.context = check_required_params(self.current_intent, self.attributes, self.context)

        # prompt being None means all parameters satisfied, perform the intent action
        if prompt is None and self.context.name!='IntentComplete':
            prompt, self.context = check_actions(self.current_intent, self.attributes, self.context)

        return prompt, self.attributes
````

**Data Loading:** Read all the utterances and extract the data (text) and labels for each intent.

````python
# YOUR CODE HERE for loading and preparing the data
import os

texts = []
labels = []

utterance_path = "/content/drive/MyDrive/Chatbot/utterances"

for file in os.listdir(utterance_path):

    if file.endswith(".dat"):

        intent_name = file.replace(".dat", "")

        with open(
            os.path.join(utterance_path, file),
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:
                sentence = line.strip()

                if sentence:
                    texts.append(sentence)
                    labels.append(intent_name)

print("Number of utterances:", len(texts))
print("Intents:", set(labels))
````

**Features extraction:** Fit the extracted text data with vectorizer to get the features.

````python
# YOUR CODE HERE to extract the features
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(ngram_range=(1, 2), lowercase=True)

X = vectorizer.fit_transform(texts)

print(X.shape)
````

**Classification:**

* Identify the features and labels
* Use train_test_split for splitting the train and test data
* Fit your model on the train set using fit() and perform prediction on the test set using predict()
* Get the accuracy of the model

        Expected Accuracy above 90%


````python
# YOUR CODE HERE for classifying the intent
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(
    X,
    labels,
    test_size=0.2,
    random_state=42
)

model = MultinomialNB()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
````

Predict the user_input using the trained model

**Note:** intentPredict() function call is specified in the Conversation Flow
- vectorize the given user_input
- reshape the vectorized array using `reshape(1,-1)` as the user_input is only a single utterance
- predict the label on the vectorized array
- return the respective class (intent_name)

````python
def intentPredict(user_input):  # Do not change the function name

    user_vector = vectorizer.transform([user_input])

    # Reshape the vectorized array as requested
    user_vector = user_vector.reshape(1, -1)

    predicted_intent = model.predict(user_vector)[0]
    print(predicted_intent)
    return predicted_intent
````

## How MultinomialNB Works — Explained With Our Chatbot Examples

---

### What is Naive Bayes?

Naive Bayes is a **probabilistic classifier** based on Bayes' Theorem:

> **P(Intent | Words) = P(Words | Intent) × P(Intent) / P(Words)**

In plain English:
> *"Given the words the user typed, what is the probability this belongs to each intent? Pick the highest one."*

It is called **Naive** because it assumes each word is **independent** of every other word — which is not always true in real language, but works extremely well in practice for short text like utterances.

It is called **Multinomial** because it works on **word counts / frequencies** (not just whether a word is present or absent).

---

### Step 1 — Training Phase (Cell 28: `classifier.fit`)

During training, the model reads all 100 utterances and learns:

| Question it answers | Example |
|---|---|
| How often does "book" appear in `book_suggestion` utterances? | Very often → high probability |
| How often does "book" appear in `get_Zodiac_Sign` utterances? | Never → very low probability |
| How often does "zodiac" appear in `book_suggestion`? | Never → very low probability |
| How often does "zodiac" appear in `get_Zodiac_Sign`? | Very often → high probability |

For each intent, the model builds a **word probability table**:

```
P(word | book_suggestion)     P(word | get_Zodiac_Sign)
─────────────────────────     ──────────────────────────
"book"      → HIGH  0.18      "book"      → LOW   ~0.001
"suggest"   → HIGH  0.12      "suggest"   → LOW   ~0.001
"novel"     → MED   0.08      "zodiac"    → HIGH  0.20
"recommend" → MED   0.07      "sign"      → HIGH  0.15
"zodiac"    → LOW   ~0.001    "born"      → MED   0.06
```

---

### Step 2 — Prediction Phase (Cell 30: `intentPredict()`)

When a user types something, the model:
1. Converts it to TF-IDF features (Cell 26)
2. Calculates the score for **each intent**
3. Returns the intent with the **highest score**

#### Example A — User types: `"suggest a book"`

```
Score(book_suggestion)  = P(book_suggestion)
                        × P("suggest" | book_suggestion)   ← HIGH
                        × P("a"       | book_suggestion)   ← medium
                        × P("book"    | book_suggestion)   ← HIGH

Score(get_Zodiac_Sign)  = P(get_Zodiac_Sign)
                        × P("suggest" | get_Zodiac_Sign)   ← LOW
                        × P("a"       | get_Zodiac_Sign)   ← medium
                        × P("book"    | get_Zodiac_Sign)   ← nearly ZERO
```

**"book" was never seen in zodiac utterances → Score(Zodiac) collapses to near zero**

✅ Result → **`book_suggestion`** (correct)

---

#### Example B — User types: `"find my zodiac sign"`

```
Score(book_suggestion)  = P(book_suggestion)
                        × P("find"   | book_suggestion)    ← medium
                        × P("zodiac" | book_suggestion)    ← nearly ZERO
                        × P("sign"   | book_suggestion)    ← nearly ZERO

Score(get_Zodiac_Sign)  = P(get_Zodiac_Sign)
                        × P("find"   | get_Zodiac_Sign)    ← medium
                        × P("zodiac" | get_Zodiac_Sign)    ← HIGH
                        × P("sign"   | get_Zodiac_Sign)    ← HIGH
```

**"zodiac" and "sign" were never seen in book utterances → Score(Book) collapses**

✅ Result → **`get_Zodiac_Sign`** (correct)

---

#### Example C — User types: `"find me a book"` (after giving author as Rowling)

After slot extraction (Cell 16), this becomes `"find me a $author"`.

The word `"$author"` is unseen — but "find", "me", "a" still tip the scale toward `book_suggestion` because those words are far more common in book utterances than zodiac ones.

✅ Result → still classified as **`book_suggestion`**

### Why TF-IDF Makes It Even Better (Cell 26)

Raw word counts have a problem — common words like **"a", "the", "I", "my"** appear in BOTH intents equally and add noise.

TF-IDF solves this by **downweighting common words** and **upweighting rare distinctive words**:

| Word | Raw count weight | TF-IDF weight | Why |
|---|---|---|---|
| `"a"` | High | LOW | Appears in every utterance — not useful |
| `"I"` | High | LOW | Same — not discriminative |
| `"book"` | Medium | HIGH | Distinctive to book_suggestion |
| `"zodiac"` | Medium | HIGH | Distinctive to get_Zodiac_Sign |
| `"suggest book"` (bigram) | Low | VERY HIGH | Only in book intent — very distinctive |

The `ngram_range=(1,2)` in Cell 26 creates **bigrams** (two-word pairs) like `"suggest book"`, `"zodiac sign"`, `"my zodiac"` — these are extremely strong signals that push accuracy to 100%.

---

### Summary — One Sentence Each

- **Training**: Count how often each word appears per intent class
- **Prediction**: Multiply those word probabilities together; pick the intent with the highest total score
- **Multinomial**: Works on word counts/frequencies (not just 0/1 presence)
- **Naive**: Treats each word independently (simplification that still works well)
- **TF-IDF + bigrams**: Removes noise from common words, amplifies intent-specific phrases
- **Result in our chatbot**: 100% accuracy — `"suggest a book"` → Book, `"zodiac sign"` → Zodiac, every time

## Cheat Sheet: Bayes Theorem -> Naive Bayes -> MultinomialNB
---

### 1. Bayes Theorem (The Foundation)

| Term | Symbol | Meaning | Our Chatbot Example |
|---|---|---|---|
| Posterior | P(Intent | Words) | Probability of an intent GIVEN the words typed | P(book_suggestion | "suggest a book") |
| Likelihood | P(Words | Intent) | How likely are these words IF we are in this intent? | P("book" | book_suggestion) = HIGH |
| Prior | P(Intent) | How common is this intent overall? | 50% - equal for both intents (50 utterances each) |
| Evidence | P(Words) | How common are these words overall? | Same for all intents - cancels out, ignored |

**Formula:** P(Intent | Words)  is proportional to  P(Words | Intent) x P(Intent)

We skip dividing by P(Words) because it is the same for all intents and cancels out.

---

### 2. Naive Bayes - The Key Assumptions

| Concept | What It Means | Why It Matters |
|---|---|---|
| Naive assumption | Every word is treated as independent of every other word | Simplifies maths - no need to calculate joint probabilities of word pairs |
| Class conditional | Word probabilities calculated per intent separately | "book" gets HIGH probability under book_suggestion, near-zero under zodiac |
| Log probabilities | Multiplications become additions | Avoids underflow when multiplying many tiny numbers |
| Laplace smoothing | Adds a small count (alpha=1) to every word, even unseen ones | Prevents probability = 0 for words not seen in training |
| Zero-probability problem | Without smoothing, one unseen word makes the whole score = 0 | With smoothing, unseen words get a tiny but non-zero probability |

---

### 3. Types of Naive Bayes - When to Use Which

| Variant | Input Type | Assumption | Best For | Used Here? |
|---|---|---|---|---|
| GaussianNB | Continuous numbers | Features follow a normal bell-curve distribution | Sensor data, numerical features | No - TF-IDF is not Gaussian |
| BernoulliNB | Binary (0 or 1) | Each feature is present or absent | Short texts with binary bag-of-words | No - ignores word frequency |
| MultinomialNB | Integer or float counts >= 0 | Features are word counts or frequencies | Text classification with TF-IDF or CountVectorizer | YES - our choice |
| ComplementNB | Integer or float counts >= 0 | Focuses on what does NOT belong to a class | Imbalanced datasets | Not needed - balanced data |

---

### 4. MultinomialNB - Step by Step

| Step | What Happens | Code In Our Notebook |
|---|---|---|
| 1. Vectorise | Convert utterance text to TF-IDF feature vector | vectorizer.fit_transform(X_train_text) - Cell 28 |
| 2. Learn word counts | For each intent, count total weighted frequency of every word | classifier.fit(X_train, y_train) - Cell 28 |
| 3. Compute P(word|intent) | word count in intent / total word count in intent + smoothing | Done internally by MultinomialNB |
| 4. Compute P(intent) | utterances for intent / total utterances | 50/100 = 0.5 for each - equal priors |
| 5. Score new input | Sum of log P(word|intent) for all words + log P(intent) | classifier.predict(X_test) - Cell 28 |
| 6. Predict | Pick the intent with the highest score | Returns book_suggestion or get_Zodiac_Sign |

---

| Metric | Value | What It Means |
|---|---|---|

| Training samples | 80 | 80% split |
| Test samples | 20 | 20% split |

### 6. MultinomialNB vs Alternatives

| Classifier | Training Speed | Data Needed | Works on TF-IDF | Verdict |
|---|---|---|---|---|
| MultinomialNB | Very fast | Very little | Perfect fit | Best choice for this task |
| Logistic Regression | Fast | Moderate | Good fit | Works but overkill |
| SVM Linear | Moderate | Moderate | Good fit | Works but slower |
| Random Forest | Slow | More data | Not ideal on sparse | Worse on sparse text |
| Neural Network | Very slow | Thousands of samples | Wasteful here | Massive overkill |
| KNN | Slow predict | Moderate | Distance not meaningful | Not suitable for text |

### 7. One-Line Answers for the Evaluator

| Question | Answer |
|---|---|
| What is Bayes Theorem? | Probability of a class given input equals likelihood of input given class multiplied by prior probability of class |
| Why Naive? | Assumes all words are independent of each other - a simplification that works very well in practice |
| Why Multinomial? | Works on word counts and frequencies (non-negative numbers), not binary presence or Gaussian distributions |
| Why not a neural network? | 100 utterances, 2 intents, distinct vocabulary - NB achieves 100% accuracy with millisecond training |
| What is Laplace smoothing? | Adding 1 to every word count so no word ever has zero probability, preventing total score collapse |
| What does TF-IDF add? | Downweights common words like a and I, upweights rare distinctive words like zodiac and suggest book |
