### Data Preprocessing

Data preprocessing is an important step in solving every machine learning problem. Most of
the datasets used with Machine Learning problems need to be processed / cleaned / transformed
so that a Machine Learning algorithm can be trained on it.

There are different steps involved in Data Preprocessing. These steps are as follows:

    1. Data Cleaning → In this step the primary focus is on
        - Handling missing data
        - Handling noisy data
        - Detection and removal of outliers
    
    2. Data Integration → This process is used when data is gathered from various data sources and data are combined to form consistent data.
    This data after performing cleaning is used for analysis.
    
    3. Data Transformation → In this step we will convert the raw data into a specified format according to the need of the model we are building.
    There are many options used for transforming the data as below:
        - Normalization
        - Aggregation
        - Generalization
        
    4. Data Reduction → Following data transformation and scaling, the redundancy within the data is removed and is organized efficiently.

## Exercise 5 - Apply Correlation Matrix (2 Marks)

Correlation is a statistical technique that can show whether and how strongly pairs of variables are related. More number of features does not imply better accuracy. More features may lead to a decline in the accuracy and create noise in the model, if they contain any irrelevant features.

*Features with high correlation value will imply the same meaning. Hence removing the highly correlated features*

**Function Description:**

`remove_Highly_Correlated()` function removes highly correlated features in the dataframe.
- Creates a correlation matrix of row and column wise features
- Extracts only uppertriangular matrix as correlation matrix, which will have the same values below and above the diagonal
- Removes columns which are having correlation value more than the threshold value.

```
def remove_Highly_Correlated(df, bar=0.9):
  # Creates correlation matrix
  corr = df.corr()

  # Set Up Mask To Hide Upper Triangle
  mask = np.triu(np.ones_like(corr, dtype=bool))
  tri_df = corr.mask(mask)

  # Finding features with correlation value more than specified threshold value (bar=0.9)
  highly_cor_col = [col for col in tri_df.columns if any(tri_df[col] > bar )]
  print("length of highly correlated columns",len(highly_cor_col))

  # Drop the highly correlated columns
  reduced_df = df.drop(highly_cor_col, axis = 1)
  print("shape of data",df.shape,"shape of reduced data",reduced_df.shape)
  return reduced_df
```

```
# YOUR CODE HERE to remove highly correlated features from the dataframe by calling above function.
reduced_df=remove_Highly_Correlated(df_merged)
```

### Optuna Optimization for BaggingClassifier

We'll use Optuna to find the best `n_estimators`, `max_samples`, and `max_features` for the `BaggingClassifier`.

```
def objective_bagging(trial):
    n_estimators = trial.suggest_int('n_estimators', 50, 300)
    max_samples = trial.suggest_float('max_samples', 0.5, 1.0)
    max_features = trial.suggest_float('max_features', 0.5, 1.0)

    model = BaggingClassifier(
        n_estimators=n_estimators,
        max_samples=max_samples,
        max_features=max_features,
        random_state=42
    )
    model.fit(X_train, y_train)
    return model.score(X_test, y_test)

print("Optimizing BaggingClassifier...")
study_bagging = optuna.create_study(direction='maximize')
study_bagging.optimize(objective_bagging, n_trials=50, show_progress_bar=True)

print("\nBest trial for BaggingClassifier:")
print(f"  Value: {study_bagging.best_value:.4f}")
print("  Params: ")
for key, value in study_bagging.best_params.items():
    print(f"    {key}: {value}")
```

### Optuna Optimization for VotingClassifier

Next, we'll optimize the `VotingClassifier`, including hyperparameters for its base estimators: `LogisticRegression`, `SVC`, and `BaggingClassifier`.

```
def objective_voting(trial):
    # Hyperparameters for LogisticRegression
    lr_C = trial.suggest_loguniform('lr_C', 1e-4, 1e4)
    lr_solver = trial.suggest_categorical('lr_solver', ['liblinear', 'lbfgs'])
    lr_model = LogisticRegression(C=lr_C, solver=lr_solver, max_iter=50000, random_state=42)

    # Hyperparameters for SVC
    svc_C = trial.suggest_loguniform('svc_C', 1e-4, 1e4)
    svc_kernel = trial.suggest_categorical('svc_kernel', ['linear', 'rbf'])
    svc_gamma = trial.suggest_categorical('svc_gamma', ['scale', 'auto'])
    svc_model = SVC(C=svc_C, kernel=svc_kernel, gamma=svc_gamma, probability=True, random_state=42)

    # Hyperparameters for BaggingClassifier (as an estimator in VotingClassifier)
    bagg_n_estimators = trial.suggest_int('bagg_n_estimators', 50, 200)
    bagg_max_samples = trial.suggest_float('bagg_max_samples', 0.5, 1.0)
    bagg_max_features = trial.suggest_float('bagg_max_features', 0.5, 1.0)
    bagg_model = BaggingClassifier(
        n_estimators=bagg_n_estimators,
        max_samples=bagg_max_samples,
        max_features=bagg_max_features,
        random_state=42
    )

    estimators = [
        ('lr', lr_model),
        ('svc', svc_model),
        ('bagg', bagg_model)
    ]

    # VotingClassifier with hard voting
    # (Soft voting requires all base estimators to support predict_proba, which they now do)
    voting_model = VotingClassifier(estimators=estimators, voting='hard')
    voting_model.fit(X_train, y_train)
    return voting_model.score(X_test, y_test)

print("\nOptimizing VotingClassifier...")
study_voting = optuna.create_study(direction='maximize')
study_voting.optimize(objective_voting, n_trials=50, show_progress_bar=True)

print("\nBest trial for VotingClassifier:")
print(f"  Value: {study_voting.best_value:.4f}")
print("  Params: ")
for key, value in study_voting.best_params.items():
    print(f"    {key}: {value}")
```

### BaggingClassifier with Optuna Optimized Parameters

Now, let's create and evaluate the `BaggingClassifier` using the best parameters found by Optuna.

```
from sklearn.ensemble import BaggingClassifier
from sklearn.metrics import classification_report

# Retrieve best parameters for BaggingClassifier
best_bagging_params = study_bagging.best_params

# Instantiate and train BaggingClassifier with optimized parameters
optimized_bagging_model = BaggingClassifier(
    n_estimators=best_bagging_params['n_estimators'],
    max_samples=best_bagging_params['max_samples'],
    max_features=best_bagging_params['max_features'],
    random_state=42
)
optimized_bagging_model.fit(X_train, y_train)

# Evaluate the optimized BaggingClassifier
y_pred_bagging = optimized_bagging_model.predict(X_test)
print("Classification Report for Optimized BaggingClassifier:")
print(classification_report(y_test, y_pred_bagging))
```

### VotingClassifier with Optuna Optimized Parameters

Next, let's create and evaluate the `VotingClassifier` using the best parameters for its components found by Optuna.

```
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import BaggingClassifier, VotingClassifier
from sklearn.metrics import classification_report

# Retrieve best parameters for VotingClassifier and its components
best_voting_params = study_voting.best_params

# Instantiate base models with optimized parameters
lr_model_opt = LogisticRegression(
    C=best_voting_params['lr_C'],
    solver=best_voting_params['lr_solver'],
    max_iter=50000,
    random_state=42
)

svc_model_opt = SVC(
    C=best_voting_params['svc_C'],
    kernel=best_voting_params['svc_kernel'],
    gamma=best_voting_params['svc_gamma'],
    probability=True,
    random_state=42
)

bagg_model_opt = BaggingClassifier(
    n_estimators=best_voting_params['bagg_n_estimators'],
    max_samples=best_voting_params['bagg_max_samples'],
    max_features=best_voting_params['bagg_max_features'],
    random_state=42
)

estimators_opt = [
    ('lr', lr_model_opt),
    ('svc', svc_model_opt),
    ('bagg', bagg_model_opt)
]

# Instantiate and train VotingClassifier with optimized parameters
optimized_voting_model = VotingClassifier(estimators=estimators_opt, voting='hard')
optimized_voting_model.fit(X_train, y_train)

# Evaluate the optimized VotingClassifier
y_pred_voting = optimized_voting_model.predict(X_test)
print("Classification Report for Optimized VotingClassifier:")
print(classification_report(y_test, y_pred_voting))
```
